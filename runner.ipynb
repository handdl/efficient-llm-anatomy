```python
# Super cool thing to understand - fused cross entnropy calculates everything in fwd!

def funny_fused_ce(grad=False, bwd=False):
    from efficient_model.loss import CrossEntropyLoss
    import torch
    from torch.utils.flop_counter import FlopCounterMode
    h = torch.randn(16,511, 512, device='cuda', dtype=torch.bfloat16, requires_grad=grad)
    w = torch.randn(16000, 512, device='cuda', dtype=torch.bfloat16, requires_grad=grad)
    lab = torch.randint(0, 16000, (16,511,), device='cuda')
    loss_fn = CrossEntropyLoss()
    with FlopCounterMode(display=False) as fc:
        loss = loss_fn(h, w, lab)
        if bwd:
            loss.sum().backward()
    print(f"{grad=}, {bwd=}: {fc.get_total_flops()/1e9:.1f} GF")
```

```python
funny_fused_ce(False, False), funny_fused_ce(True, False), funny_fused_ce(True, True)
```
